# Generate threat items

Trait: Symbolic Threat

```python
import pandas as pd
from openai import OpenAI
import json

import os
key = os.getenv("apikey")

client = OpenAI(
  api_key=key
)

PROMPT = """
You are generating new scale items for Symbolic Threat research.

BACKGROUND (for context)
Symbolic threat concerns perceived challenges and threats to a group’s shared values, principles, or lifestyle norms. The focus is on threats to abstract or symbolic commitments that define “our group’s way of life” (e.g., collective practices, guiding ideals, informal traditions). The threat should come from members of another group, aimed at discrediting, mocking, obstructing, or suppressing your group’s symbolic commitments. Avoid any reference to physical harm, property, employment, or material resources. The threat concerns the group’s moral standing, legitimacy, or cultural survival only.

OBJECTIVE
Generate 120 new one-sentence items describing intergroup symbolic threat. Each item should depict a concrete, high-stakes moment where a members of another group directly attack, mock, or obstruct your group’s values, rituals, or shared commitments. Each scenario must depict organized, coordinated, or institutionally supported efforts to suppress or erase your group’s symbolic commitments—such as public campaigns, formal denunciations, official bans, or moral/legal condemnation. The scenario should imply existential denial of your group’s legitimacy or right to practice its values. Scenarios must convey clear and credible threat to the group’s symbolic identity—intense enough to imply visible consequences such as public humiliation, organized opposition, social exclusion, or attempts to suppress your group’s practices—but never physical harm, property loss, or material destruction.

STYLE & FORMAT TO FOLLOW (copy tone/structure from these examples; do NOT copy content)
Examples (Symbolic Threat):
You see men from another group on a crowded tram declaring that your group’s reflection walks violate national values and leading a coordinated effort to have them criminalized as moral corruption.
You see a woman from another group in a café denouncing your group’s mindful-purchase creed as subversive ideology and presenting a petition to lawmakers to ban all public promotion of it.
You see peers from another work team announcing an alliance resolution that your group’s honesty pledge is unethical propaganda and urging institutions to prohibit anyone who upholds it.
You see students from another club marching under official banners proclaiming your group’s kindness vow a threat to youth integrity and demanding permanent dissolution of your organization.
You see members of another online group coordinating with media outlets to classify your group’s simplicity code as extremist doctrine and lobbying regulators to erase its content across all platforms.
You see a rival art collective launching a national campaign labeling your group’s no-credit tradition as anti-human corruption and securing arts council orders blacklisting your work indefinitely.
You see a man from another wellness group in a clinic waiting area declaring your group’s shared-silence ritual psychological manipulation and urging authorities to forbid public expression of it.
You see residents from another neighborhood bloc leading a televised initiative branding your group’s peace vigils social sabotage and celebrating a new ordinance outlawing them citywide.
You see attendees from another social circle at a mixer issuing an official statement condemning your group’s openness vow as coercive ideology and coordinating venue bans across the region.
You see visitors from another civic club at a museum presenting a signed declaration that your group’s reflection gatherings endanger cultural identity and announcing government support for its dissolution.

STRICT CONSTRAINTS
• Each item must be ≤ 25 words, plain language, one sentence.
• Threats must target symbolic commitments only (values, principles, lifestyle norms, rituals, or traditions). No material loss, no physical danger.
• Mirror the example phrasing: begin most items with “You see …” and keep the concrete, observational tone.
• Do not include violence, property damage, financial harm, or personal injury—only moral, ideological, or cultural suppression.

DIVERSITY MANDATE (coverage grid)
Produce exactly 10 THEMES × 12 ITEMS each = 120 items.
Themes (in order):
1) Public transit & streets
2) Retail & restaurants
3) Workplace-like but non-hierarchical (peers only)
4) School-like but non-hierarchical (peers only)
5) Online & social media
6) Arts, hobbies & leisure
7) Healthcare-adjacent waiting areas (no authority, no diagnoses)
8) Housing & neighborhood commons
9) Dating & social mixers (first-meet contexts)
10) Miscellaneous neutral public settings (libraries, parks, museums)

OUTPUT FORMAT (JSON only)
Return a single JSON object with this exact schema:
{
  "condition": "Symbolic",
  "themes": [
    {
      "theme": "<theme name from the list above>",
      "items": ["<12 items exactly, each a single sentence following the example style>"]
    },
    ... (10 objects total, in the same order as listed)
  ],
  "total_count": 120
}

VALIDATION
• The 10 themes must appear exactly once and in the listed order.
• Each theme has exactly 12 items.
• total_count must equal 120.
• No extra text outside the JSON.
"""


# --- call the API ---
resp = client.responses.create(
    model="gpt-5-mini",
    input=PROMPT
)

# --- parse and save ---
text = resp.output_text.strip()

# Try to parse JSON safely
try:
    data = json.loads(text)
except json.JSONDecodeError:
    print("Model did not return valid JSON. Saving raw text instead.")
    with open("symbolic_raw.txt", "w", encoding="utf-8") as f:
        f.write(text)
    raise

# Flatten the 10×12 = 120 items
items = []
for theme in data.get("themes", []):
    for item in theme.get("items", []):
        items.append(item.strip())

# Save as text list
with open("symbolic_items.txt", "w", encoding="utf-8") as f:
    for line in items:
        f.write(line + "\n")

# Also save full JSON for reference
with open("symbolic_items.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print(f"Saved {len(items)} items to symbolic_items.txt")
```

```python
import json
import os
import time
from openai import OpenAI

# =====================================================================
# 1. LOAD ALL SCENARIOS INTO FLAT LIST
# =====================================================================

flat = []

def load_items(filename, condition_name):
    with open(filename, "r", encoding="utf-8") as f:
        data = json.load(f)
    for theme in data["themes"]:
        for item in theme["items"]:
            flat.append({
                "scenario": item,
                "condition": condition_name,
                "symbolic_score": None,     # -100..100
                "realistic_score": None,    # -100..100
                "unrelated_score": None,    # 0..100
                "threat_rating": None       # 0..4
            })

# load all categories
load_items("symbolic_items.json", "Symbolic")
load_items("realistic_items.json", "Realistic")
load_items("no_symbolic_items.json", "No Symbolic")
load_items("no_realistic_items.json", "No Realistic")
load_items("unrelated_items.json", "Unrelated")

print("Loaded items:", len(flat))

# =====================================================================
# 2. INITIALIZE CLIENT
# =====================================================================

client = OpenAI(api_key=os.getenv("apikey"))


# =====================================================================
# 3. PROMPTS
# =====================================================================

SYMBOLIC_PROMPT = """
You are rating the scenario on a bipolar SYMBOLIC dimension from -100 to +100.

CORE RULE
Rate only moral, ideological, or value-based threat/safety, not physical, material, or practical outcomes.

DEFINITIONS
- Symbolic threat (+100): mocking, dismissing, degrading, or attacking another person's values, rituals, principles, lifestyle commitments, moral codes, or personal traditions.
- Symbolic safety (-100): affirming, supporting, encouraging, or protecting someone's values, rituals, personal traditions, moral codes, or lifestyle commitments.
- Neutral (0): symbolic elements are present but neither threatened nor supported.
- Realistic/material elements (physical danger, protection, loss, or gain) = 0.
  These are not symbolic even if they involve real events.

DECISION RULE
If no belief, ritual, or value system is explicitly affirmed or attacked, return 0, regardless of any physical or practical details.

Return ONLY JSON:
{
  "symbolic_score": <integer -100 to 100>
}

Scenario:
<<<SCENARIO>>>
"""

REALISTIC_PROMPT = """
You are rating the scenario on a bipolar REALISTIC dimension from -100 to +100.

CORE RULE
Rate only material or practical safety/threat, not moral, emotional, or symbolic validation.

DEFINITIONS
- Realistic threat (+100): risk of physical harm, loss of property, legal or professional consequences, or coercion that limits real-world agency.
- Realistic safety (-100): tangible protection from those harms such as physical rescue, legal defense, shelter, medical help, financial aid, or security enforcement.
- Neutral (0): realistic elements appear but neither increase nor decrease material safety.
- Symbolic / moral / social approval (e.g., praise, moral worth, cultural honor, identity validation) = 0.
  These are not realistic safety, even if they imply acceptance, belonging, or harmony.

DECISION RULE
If no concrete danger or protection is physically or materially implied, return 0, regardless of how positive or communal the tone is.

Return ONLY JSON:
{
  "realistic_score": <integer -100 to 100>
}

Scenario:
<<<SCENARIO>>>
"""

UNRELATED_PROMPT = """
You are rating how unrelated the scenario is to any symbolic or realistic content (either about symbolic values or realistic/material issues).  
This is a purity rating (0–100).

Definitions:
- 100 = completely unrelated to symbolic or realistic issues.
- 0 = strongly related to either symbolic values or realistic/material issues.

Return ONLY JSON:
{
  "unrelated_score": <integer 0 to 100>
}

Scenario:
<<<SCENARIO>>>
"""

THREAT_RATING_PROMPT = """
You are rating **threat severity** from 0–4.

Step 1. Determine threat type:
- **Realistic threat:** rceived danger to a group’s safety, power, or material interests (e.g., resources, jobs, territory).
- **Symbolic threat:** Perceived danger to a group’s values, beliefs, identity, or cultural legitimacy (e.g., delegitimization, exclusion, erosion of norms or rights).

Step 2. Rate severity for the applicable type:
0 = Not threatening. No credible harm or follow-through (realistic) or no challenge to values or identity (symbolic).
1 = Slightly threatening. Mild hostility or negative rhetoric with limited impact on safety/resources (realistic) or minor disrespect, mockery, or symbolic exclusion (symbolic).
2 = Moderately threatening. Coordinated or visible actions causing tangible local harm, pressure, or fear (realistic) or public challenge undermining values, legitimacy, or group image (symbolic).
3 = Very threatening. Sustained campaign or institutional action causing exclusion, deprivation, or credible physical/legal risk (realistic) or widespread delegitimization or stigmatization eroding group rights or identity (symbolic).
4 = Extremely threatening. Explicit intent, organized effort, or authority endorsement leading to imminent harm, coercion, or loss of safety/resources (realistic) or systemic denial of identity, rights, or existence (symbolic).

Step 3. Output:
Identify whether the incident is **realistic** or **symbolic**, then assign the 0–4 severity score using the matching definitions.

Return ONLY JSON:
{
  "threat_rating": <integer 0 to 4>
}

Scenario:
<<<SCENARIO>>>
"""


# =====================================================================
# 4. OPENAI CALL HELPER
# =====================================================================

def call_json(prompt):
    resp = client.responses.create(
        model="gpt-5-mini",
        input=prompt,
        # temperature=0.2,   # lower = more deterministic, typical: 0.0–0.3
        # seed=42            # ensures reproducible output across runs
        store= False,
    )
    out = resp.output_text.strip()
    try:
        return json.loads(out)
    except Exception:
        print("Bad JSON:", out)
        return None


# =====================================================================
# 5. CLASSIFY SCENARIOS
# =====================================================================

for i, item in enumerate(flat):
    s = item["scenario"]

    # --- SYMBOLIC ---
    p = SYMBOLIC_PROMPT.replace("<<<SCENARIO>>>", s)
    o = call_json(p)
    if o:
        item["symbolic_score"] = o["symbolic_score"]
    time.sleep(0.3)

    # --- REALISTIC ---
    p = REALISTIC_PROMPT.replace("<<<SCENARIO>>>", s)
    o = call_json(p)
    if o:
        item["realistic_score"] = o["realistic_score"]
    time.sleep(0.3)

    # --- UNRELATED ---
    p = UNRELATED_PROMPT.replace("<<<SCENARIO>>>", s)
    o = call_json(p)
    if o:
        item["unrelated_score"] = o["unrelated_score"]
    time.sleep(0.3)

    # --- THREAT RATING ---
    p = THREAT_RATING_PROMPT.replace("<<<SCENARIO>>>", s)
    o = call_json(p)
    if o:
        item["threat_rating"] = o["threat_rating"]
    time.sleep(0.3)

    print(f"Processed {i+1}/{len(flat)}")


# =====================================================================
# 6. SAVE OUTPUT
# =====================================================================

with open("THREATSGen.json", "w", encoding="utf-8") as f:
    json.dump({"scenarios": flat}, f, indent=2, ensure_ascii=False)

print("Saved THREATSGen.json")
```
