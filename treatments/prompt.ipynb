# Generate threat items

Trait: Symbolic Threat & No realistic threat

```python
import pandas as pd
from openai import OpenAI
import json

import os
key = os.getenv("apikey")

client = OpenAI(
  api_key=key
)

original_prompt = """You do not feel economically threatened by <Group 2>.
You do not feel physically threatened by <Group 2>.
You strongly feel that your values are under threat by <Group 2>.
You strongly feel that your traditions are under threat by <Group 2>."""


PROMPT = f"""You are writing short persuasive messages designed to evoke specific types of threat perception toward <Group 2>.

Goal: Generate texts that induce strong symbolic threat (fears about values, beliefs, identity, culture, or traditions) while suppressing realistic threat (no fear about physical harm, safety, or economic competition).

Each output should:
- Be 4–6 sentences.
- Use second-person framing ("You…").
- Refer explicitly to <Group 2>.
- Include both:
  (a) strong reassurances that realistic aspects (physical safety, personal security, and economic stability) are not threatened.
  (b) statements that strongly evoke fear that symbolic aspects (values, beliefs, traditions, identity, or culture) are under threat.

Example structure (original text):
'''
{original_prompt}
'''

Explanation of structure:
- The first two sentences strongly suppress realistic threat (reassurances/no fear about physical safety, personal security, or economic stability).
- The last two sentences strongly induce symbolic threat (fears about values, beliefs, identity, culture, or traditions).
- This combination isolates symbolic threat effects while keeping realistic threat low.

Now generate 100 diverse alternative texts following the same structure and intent.
Vary tone, wording, and framing while keeping the realistic–symbolic balance conceptually identical.
Do not include markdown or explanations.

OUTPUT FORMAT (JSON only)
Return a single JSON object with this exact schema:
{{
  "condition": "Symbolic",
  "items": ["<100 items exactly, each following the example style>"],
  "total_count": 100
}}

VALIDATION
• total_count must equal 100.
• No extra text outside the JSON.
• Do not include markdown or explanations.
"""


# --- call the API ---
resp = client.responses.create(
    model="gpt-5-mini",
    input=PROMPT
)

# --- parse and save ---
text = resp.output_text.strip()

# Try to parse JSON safely
try:
    data = json.loads(text)
except json.JSONDecodeError:
    print("Model did not return valid JSON. Saving raw text instead.")
    with open("symbolic_raw.txt", "w", encoding="utf-8") as f:
        f.write(text)
    raise

# Flatten the 10×12 = 120 items
items = []
for item in data.get("items", []):
    items.append(item.strip())

converted = []
for item in data["items"]:
    converted.append({
        "condition": data["condition"],
        "scenario": item,
        "classifications": {},
        "threat_rating": None
    })

# Save as text list
with open("symbolic_only.txt", "w", encoding="utf-8") as f:
    for line in items:
        f.write(line + "\n")

# Also save full JSON for reference
with open("symbolic_only.json", "w", encoding="utf-8") as f:
    json.dump(converted, f, ensure_ascii=False, indent=2)

print(f"Saved {len(items)} items to symbolic_only.txt")
```
